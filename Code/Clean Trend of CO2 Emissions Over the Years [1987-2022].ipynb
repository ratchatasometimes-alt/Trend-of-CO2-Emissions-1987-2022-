```python
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
df = pd.read_csv(r'C:\Users\User\Downloads\archive (7).zip')
```

```python
print("จำนวนค่าที่หายไปในแต่ละคอลัมน์:")
print(df.isna().sum())
print("\nข้อมูลหลังจากลบแถวที่มีค่าที่หายไป:")
print(df.dropna())
print("\nข้อมูลหลังจากแทนค่าที่หายไปด้วยค่าเฉลี่ยของคอลัมน์:")
print(df.fillna(df.mean(numeric_only=True)))
```

```python
print("ข้อมูลสรุปของ DataFrame:")
print(df.info())
print("\nตัวอย่างข้อมูล:")
print(df.head())
print("\nจำนวนแถวและคอลัมน์ของข้อมูล:")
print(df.shape)
```

```python
print("ค่าเฉลี่ยของแต่ละคอลัมน์:")
print(df.mean(numeric_only=True))
print("\nส่วนเบี่ยงเบนมาตรฐานของแต่ละคอลัมน์:")
print(df.std(ddof=1, numeric_only=True))
print("\nสถิติสรุปของแต่ละคอลัมน์:")
print(df.describe())
```
